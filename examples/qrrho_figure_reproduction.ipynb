```python
# TODO: Write up the significance of this reproduction in markdown cells

from JDFTxFreeNrg.qrrho import get_qrrho_vib_entropies, get_qrrho_vib_enthalpies
from JDFTxFreeNrg.standard import get_enthalpy_vib, get_entropy_vib, k_ev
from scipy.constants import eV, N_A
import numpy as np
import matplotlib.pyplot as plt
eV_to_J = eV * N_A  # eV to J / molecule

test_freqs = np.arange(0.1, 400, 0.01)
T = 300.
```

```python
qrrho_vib_enthalpies = get_qrrho_vib_enthalpies(test_freqs, T=T)
rrho_vib_enthalpies = get_enthalpy_vib(test_freqs, T=T)

plt.plot(test_freqs, qrrho_vib_enthalpies, label='QRRHO Vib Enthalpy', color="green")
plt.axhline(k_ev*T/2, color="green", linestyle='--', label='QRRHO Zero-Freq Limit')
plt.plot(test_freqs, rrho_vib_enthalpies, label='RRHO Vib Enthalpy', color="red")
plt.axhline(k_ev*T, color="red", linestyle='--', label='RRHO Zero-Freq Limit')
plt.xlabel('Frequency (cm$^{-1}$)')
plt.ylabel('Vibrational Enthalpy (eV)')
plt.legend()
```

```python
qrrho_vib_entropies = get_qrrho_vib_entropies(test_freqs, T=T)
rrho_vib_entropies = get_entropy_vib(test_freqs, T=T)

fig, ax = plt.subplots()
twinx = ax.twinx()
ax.plot(test_freqs, qrrho_vib_entropies, label='QRRHO Vib Entropy', color="green")
twinx.plot(test_freqs, qrrho_vib_entropies * eV_to_J, color="green")
ax.plot(test_freqs, rrho_vib_entropies, label='RRHO Vib Entropy', color="red")
twinx.plot(test_freqs, rrho_vib_entropies * eV_to_J, color="red")
ax.set_xlabel('Frequency (cm$^{-1}$)')
ax.set_ylabel('Vibrational Entropy (eV / K)')
twinx.set_ylabel('Vibrational Entropy (J / mol K)')
ax.legend()

```
