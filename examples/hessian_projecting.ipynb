# Intro

```python
from os import getcwd
from pathlib import Path
from pymatgen.io.jdftx.inputs import JDFTXInfile

from JDFTxFreeNrg.hessian import read_K, solve_vib_modes, get_freqs, freq_nrg_to_cm, print_freqs

hydronium_calc_dir = Path(getcwd()) / "data" / "solvated_hydronium"
molecule_sets = None

infile = JDFTXInfile.from_file(hydronium_calc_dir / "in")
structure = infile.structure
K_raw = read_K(hydronium_calc_dir, structure)

omegaSqEigs, omegaEvecs = solve_vib_modes(structure, K_raw)
freqs_Ha = get_freqs(omegaSqEigs)
freqs_cm = freq_nrg_to_cm(freqs_Ha)
print_freqs(freqs_cm)
```

# Oh no, 4 im freqs. Project out external DOFs

```python
from JDFTxFreeNrg.hessian import get_projected_K
molecule_sets = [
    {
        "indices": "all",
        "rot": True,
        "axes": None,
    },
    {
        "indices": "all",
        "trans": True,
        "axes": None,
    },
]
K_proj = get_projected_K(structure, K_raw, molecule_sets=molecule_sets)
omegaSqEigs, omegaEvecs = solve_vib_modes(structure, K_proj)
freqs_Ha = get_freqs(omegaSqEigs)
freqs_cm = freq_nrg_to_cm(freqs_Ha)
print_freqs(freqs_cm)
```
